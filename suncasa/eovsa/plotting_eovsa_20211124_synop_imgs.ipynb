```python
import matplotlib
%matplotlib inline
from sunpy import map as smap
import os   
from glob import glob
import matplotlib.pyplot as plt
import math
import warnings
warnings.filterwarnings("ignore", category=UserWarning, module="sunpy")
```

```python
workdir='/Users/fisher/myworkspace_local/temp_20211124'
os.chdir(workdir)
```

```python
spwstrs = ['00-01','02-04','05-10','21-30','31-40','41-49']
spwstr = spwstrs[2]
eofiles = sorted(glob(f'eovsa_20211124T??????_*min.s{spwstr}.init.tb.fits'))
# eofiles = sorted(glob(f'eovsa_20211124T??????_*min.s{spwstr}.tb.fits'))

print(eofiles)
nimgs = len(eofiles)
# fig = plt.figure(figsize=(nimgs*3, 3))
# Calculate the number of rows and columns
ncols = math.ceil(math.sqrt(nimgs))
nrows = math.ceil(nimgs / ncols)

fig = plt.figure(figsize=(ncols * 3, nrows * 3))
# titles=['original','corrected']
for fidx, file in enumerate(eofiles):
    eomap = smap.Map(file)
    ax = fig.add_subplot(nrows, ncols, fidx + 1, projection=eomap)
    eomap.plot(axes=ax, cmap='jet')
fig.tight_layout()
```
